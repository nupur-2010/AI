TEXT FILES

1) Create Directory

```python
import os

os.makedirs("Data/TextFiles",exist_ok=True)
```

2. Create Text Files in Directory

```python
sample_text = {
    "Data/TextFiles/AI.txt" : """ Artificial Intelligence (AI) is the broader field of computer science focused on creating computer systems that can perform tasks that normally require human intelligence. 
These tasks include learning from experience, reasoning, problem solving, understanding natural language, recognizing images and speech, making decisions, and interacting with humans. 
AI includes many different approaches and technologies, including rule-based systems, Machine Learning, Deep Learning, Natural Language Processing, computer vision, robotics, and Generative AI.
Machine Learning and Deep Learning are therefore important parts of the broader field of Artificial Intelligence.
""",
    "Data/TextFiles/ML.txt" : """ Machine Learning (ML) is a subset of Artificial Intelligence that enables computers to learn patterns and relationships from data and make predictions or decisions without being explicitly programmed for every individual task. 
Instead of relying only on predefined rules, machine learning algorithms learn from examples and improve their performance as they are exposed to more data. 
Machine Learning is commonly divided into supervised learning, unsupervised learning, and reinforcement learning. 
It is widely used in applications such as spam detection, fraud detection, recommendation systems, customer segmentation, price prediction, and forecasting.
""",
    "Data/TextFiles/DL.txt" : """ Deep Learning (DL) is a specialized subset of Machine Learning that uses artificial neural networks with multiple layers to automatically learn complex patterns and representations from large amounts of data. 
Unlike many traditional machine learning approaches, deep learning can automatically learn useful features from raw data, reducing the need for extensive manual feature engineering. 
Deep Learning is particularly powerful for working with images, audio, video, and natural language.
Popular deep learning architectures include Convolutional Neural Networks (CNNs), Recurrent Neural Networks (RNNs), Transformers, and other neural network models.
"""
}

for filepath, content in sample_text.items():
    with open(filepath, "w", encoding="utf-8") as file:
        file.write(content.strip())

print("Sample text files created successfully.")
```

3. Load single text file from the directory

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("Data/TextFiles/AI.txt",encoding="utf-8")
document = loader.load()
document
```

4. Load all text files from the directory

```python
from langchain_community.document_loaders import DirectoryLoader, TextLoader

loader = DirectoryLoader(
    "Data/TextFiles",
    glob="**/*.txt",
    loader_cls=TextLoader,
    loader_kwargs={"encoding" : "utf-8"},
    show_progress=True
)

documents = loader.load()
documents
```

PDF FILES

```python
from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader, PyMuPDFLoader

loader = DirectoryLoader(
    "Data/PdfFiles",
    glob="**/*.pdf",
    loader_cls=PyMuPDFLoader,
    show_progress=True
)

documents = loader.load()
documents
```
